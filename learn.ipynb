```python
has_python = df['LanguageHaveWorkedWith'].str.contains(r'\bPython\b', regex=True, na=False)
python_by_country = has_python.groupby(df['Country']).sum()
python_by_country
```

```python
country_res = df['Country'].value_counts()
country_res
```

```python
python_df = pd.concat([country_res, python_by_country], axis='columns', sort=False)
python_df
```

```python
python_df.rename(columns={'LanguageHaveWorkedWith': 'Python users'}, inplace=True)
```

```python
python_df['knowPythonPercent'] = (python_df['Python users']/python_df['count']) * 100
python_df
```
